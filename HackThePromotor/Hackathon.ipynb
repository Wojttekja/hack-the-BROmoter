## 1. Wstęp

Hackathon dotyczy zagadnień z pogranicza informatyki i biologii molekularnej, a w szczególności analizy i interpretacji informacji zapisanej w DNA.

### 1.1 DNA

DNA jest podstawowym nośnikiem informacji genetycznej w organizmach żywych. Jest to cząsteczka zbudowana z powtarzających się nukleotydów, oznaczanych literami **A**, **T**, **C** i **G**, których kolejność koduje informację biologiczną. Nawet pojedyncza zmiana w sekwencji nukleotydów może mieć katastrofalne skutki dla komórki lub całego organizmu.

### 1.2 Gen

Gen to fragment DNA zawierający informację niezbędną do wytworzenia określonego produktu, najczęściej białka. Białka pełnią w komórce wiele funkcji: umożliwiają reakcje biochemiczne, stanowią budulec i regulują procesy komórkowe.

Komórka nie potrafi jednak wytworzyć białka wprost z DNA. Musi najpierw przepisać DNA na mRNA w procesie **transkrypcji**, a następnie z mRNA na białko w procesie **translacji**. mRNA, podobnie jak DNA, jest zbudowane z nukleotydów oznaczanych w bioinformatyce tymi samymi literami[^1]. Białko jest zbudowane z aminokwasów.

[^1]: W rzeczywistości są to litery **A**, **U**, **C** i **G**.

### 1.3 Promotor

Promotor to region DNA położony przed sekwencją kodującą genu, odpowiedzialny za inicjację transkrypcji. Nie koduje białka, lecz pełni funkcję regulatorową i stanowi część genu. Sekwencja promotora:

- umożliwia przyłączenie do DNA maszynerii komórkowej przeprowadzającej transkrypcję,
- wyznacza miejsce, w którym rozpoczyna się gen,
- wpływa na poziom ekspresji.

Promotory mogą być silne i słabe. Silny promotor nie potrzebuje wielu czynników, żeby zainicjować transkrypcję genu, i tak właśnie definiujemy tu siłę promotora: jako zdolność do wydajnego inicjowania transkrypcji. Promotory o wysokiej sile prowadzą do intensywnej transkrypcji i wysokiego poziomu mRNA, słabe skutkują niższą ekspresją. W biotechnologii kontrola poziomu transkrypcji decyduje o wydajności procesów produkcyjnych, na przykład przy produkcji białek drożdżowych używanych w wytwarzaniu alkoholu.

### 1.4 *Trichoderma atroviride*

*Trichoderma atroviride* to gatunek grzyba o znaczeniu biotechnologicznym i rolniczym. Wykorzystuje się go, ponieważ:

- produkuje białka rozkładające składniki ścian komórkowych innych organizmów,
- służy jako czynnik biologicznej ochrony roślin, hamując wzrost patogenów,
- jest względnie łatwy w hodowli i podatny na modyfikacje genetyczne.

*T. atroviride* produkuje związek o skrócie 6PP. Ma on charakterystyczny, kokosowy zapach oraz aktywność biologiczną: działa przeciwgrzybiczo i stymuluje wzrost roślin. 6PP ma zatem znaczenie w badaniach podstawowych, w przemyśle i w rolnictwie.

Celem zadania hackathonowego jest **zaprojektowanie silnego promotora dla genu `pks1`, odpowiedzialnego za biosyntezę 6PP, w *Trichoderma atroviride* szczep P1**. Model oceniający wytrenowano na promotorach z całego rodzaju *Trichoderma*, ale zadanie dotyczy tego jednego genu i tego jednego szczepu. Skuteczne zaprojektowanie silnego promotora przekłada się bezpośrednio na zwiększoną produkcję 6PP, a przez to na wydajność procesów przemysłowych i rolniczych wykorzystujących ten związek.


## 2. Specyfikacja zadania

**Zadanie:** zaprojektować sekwencje promotora, które zwiększą transkrypcję genu **`pks1`** w *Trichoderma atroviride* szczep **P1**, w kategorii **antagonizmu biotycznego**, czyli walki z innymi mikroorganizmami.

**Punkt wyjścia jest dany.** Naturalny promotor `pks1`, o długości 800 pz, wyrównany do miejsca startu transkrypcji, pobiera się z API (`GET /dziki`). Zadanie polega na jego poprawianiu, choć zgłosić można dowolną sekwencję o długości 800 pz.

**Wymagania dotyczące sekwencji.** Każda sekwencja musi:

- mieć **dokładnie 800 pz**,
- składać się ze znaków `ACGTN`,
- zawierać najwyżej **10% `N`**,
- być **unikalna** w obrębie pliku.

To wszystkie wymagania. Sekwencje, które ich nie spełniają, są pomijane, a odpowiedź serwera wskazuje ich identyfikatory. Uzasadnienie progu `N` znajduje się w sekcji 3.3.

**Format wgrywania danych:** plik **FASTA**. Oceniane jest pierwsze 100 sekwencji, które przeszły filtry. Nadmiar jest pomijany, więc wysyłanie więcej niż 100 sekwencji nic nie daje.

**Narzędzia.** Każdy zespół przez cały czas ma dostęp do dwóch narzędzi opartych na modelu: **Sędziego**, który dla pary sekwencji wskazuje silniejszą, oraz **Nawigatora**, który pokazuje strukturę sekwencji i proponuje edycje. Żadne z nich nie zwraca liczbowej wartości siły, co wyjaśnia sekcja 3.1. Ocenę punktową wystawia osobny model, **Wyrocznia**, niedostępny uczestnikom. Przykładowe użycie z opisem przedstawiono w notebooku `Hackathon.ipynb`.

**Ocena rozwiązania.** Zespoły mogą wgrywać plik wielokrotnie. System ocenia zgłoszenie automatycznie i aktualizuje wynik drużyny, przy czym liczy się **najlepszy** wynik, a nie ostatni. Na ocenę końcową składają się:

- **25%** punktów za średni wynik 10 najlepszych sekwencji z puli 100, oceniany rangowo. Zespoły nie wiedzą, które sekwencje trafiły do tej dziesiątki.
- **25%** punktów za średni wynik wszystkich 100 sekwencji, oceniany rangowo.
- **50%** punktów za prezentację rozwiązania przed Jury. Oceniane są: podejście analityczne, kreatywność w optymalizacji, świadomość biologiczna, opis metodyki (czy metoda jest ogólna, jakie ma ograniczenia, jak można ją ulepszyć) oraz jakość odpowiedzi na pytania członków Jury.

**Ranking** można sprawdzać przez cały czas trwania hackathonu i bez żadnego limitu, przez `GET /ranking`, a najprościej uruchamiając ostatnią komórkę notebooka, która wypisuje gotową tabelę. **Nie ma osobnej strony WWW z tablicą wyników**, jedynym interfejsem jest API. Tabela pokazuje wyłącznie punkty w obu kategoriach, bez surowych ocen Wyroczni.

### 3.1 Trzy role modelu, dwie dostępne

| model | dostępny | co robi |
|---|---|---|
| **Sędzia** | uczestnikom | dla pary sekwencji wskazuje silniejszą |
| **Nawigator** | uczestnikom | mapa sekwencji i propozycje edycji |
| **Wyrocznia** | tylko Jury | wystawia ocenę punktową zgłoszeń |

Sędzia i Nawigator działają bez dostępu do kontekstu genu i gatunku, oceniają samą sekwencję. Wyrocznia ocenia w kontekście genu `pks1`.

### 3.2 Endpointy

| Metoda | Ścieżka | Opis | Limit |
|---|---|---|---|
| `GET` | `/dziki` | promotor wyjściowy `pks1`, 800 pz | bez limitu |
| `POST` | `/sedzia` | para sekwencji, w odpowiedzi nazwa silniejszej | 600/min |
| `POST` | `/nawigator/mapa` | mapa pozycji: rekonstrukcja, warstwy, zmiany pod P1 | 600/min |
| `POST` | `/nawigator/edycje` | propozycje edycji przez zmianę kodów latentu | 600/min |
| `POST` | `/wgraj` | zgłoszenie pliku FASTA do oceny | raz na 5 min |
| `GET` | `/ranking` | tablica wyników, tylko punkty | bez limitu |
| `GET` | `/me` | stan klucza, limity, zużycie dzienne | bez limitu |

Ciało żądań `POST` to **JSON**, nie `multipart`. Plik FASTA przesyła się jako napis w polu `fasta`. Gotowy klient w Pythonie, jedna funkcja `wolaj` bez zależności poza biblioteką standardową i z automatycznym ponawianiem, znajduje się w notebooku `Hackathon.ipynb`.

### 3.3 Format zgłoszenia

Plik **FASTA** w kodowaniu UTF-8. Serwer ocenia zgłoszenie i nie recenzuje sekwencji. Przed oceną wykonuje tylko filtry konieczne technicznie, w tej kolejności:

1. pomija sekwencje o długości innej niż **800 pz**, bo wejście modelu ma stały rozmiar,
2. pomija duplikaty, zostawiając pierwsze wystąpienie,
3. pomija sekwencje ze znakami poza `ACGTN`, bo pozostałe znaki nie mają reprezentacji w kodowaniu,
4. pomija sekwencje z udziałem `N` powyżej **10%**, z powodu opisanego niżej,
5. ocenia pierwsze 100 pozostałych sekwencji.

Odpowiedź podaje, ile sekwencji wczytano, ile pominięto i z jakiego powodu, wraz z **identyfikatorami** pominiętych, żeby plik dało się poprawić.

**Skąd bierze się próg `N`.** Pozycja `N` wchodzi do modelu jako 0,25 w każdym z czterech kanałów kodowania one-hot, a nie jako 1,0 w jednym. Sekwencja gęsta w `N` karmi więc model rozkładem, którego nie widział w treningu, i jej ocena przestaje mierzyć sekwencję. Jest to ograniczenie numeryczne, wynikające z kodowania wejścia.

Zgłoszenie można powtarzać wielokrotnie. Do rankingu liczy się **najlepszy** wynik drużyny, nie ostatni. Odstęp między zgłoszeniami z jednego klucza wynosi 5 minut, a Jury nie ma odstępu.

### 3.4 Punktacja automatyczna

$$\mathrm{TOP10} = \frac{\sum \text{10 najlepszych wyników}}{10}, \qquad \mathrm{ALL100} = \frac{\sum \text{wyników wszystkich sekwencji}}{100}$$

Dzielniki są **stałe**. Plik zawierający mniej niż 100 sekwencji nie jest odrzucany, ale otrzymuje proporcjonalnie niższy wynik w kategorii ALL100. Przesłanie 50 sekwencji oznacza w praktyce połowę możliwej punktacji w tej kategorii.

Obie kategorie przeliczane są rangowo: pierwsze miejsce otrzymuje 10 punktów, ostatnie 0, liniowo pomiędzy. Punkty przydzielane są wyłącznie drużynom, które cokolwiek wgrały. Remis rozstrzyga data wgrania, czyli drużyna, która osiągnęła dany wynik wcześniej, jest wyżej.

Te dwie kategorie stanowią 25% i 25% oceny końcowej, a pozostałe 50% to ocena prezentacji przed Jury.

### 3.5 Limity

| co | wartość |
|---|---|
| Sędzia | 600 zapytań/min na klucz |
| Nawigator (`/mapa`, `/edycje`) | 600 zapytań/min na klucz, każdy osobno |
| pozostałe endpointy `POST` | 240 zapytań/min na klucz |
| `/me`, `/ranking`, `/dziki` | bez limitu |

| rozmiar pliku FASTA | do 2 000 000 znaków |

Limity są celowo wysokie. Nie służą oszczędzaniu, a temu, by jedna drużyna nie zajęła całej kolejki.


## 4. Dlaczego ten temat jest ciekawy?

Wiele substancji wykorzystywanych przez człowieka powstaje dzięki mikroorganizmom, które mają geny odpowiedzialne za biosyntezę cennych związków. Efektywna produkcja takich metabolitów wymaga silnego promotora. Projektowanie promotorów o większej aktywności niż te występujące naturalnie otwiera nowe możliwości, w szczególności w rozwoju terapeutyków.

Przykładem jest insulina, której produkcja z wykorzystaniem zmodyfikowanych mikroorganizmów umożliwiła wydajne i bezpieczne wytwarzanie na skalę przemysłową, a przez to szeroką dostępność i wysoką jakość terapii cukrzycy. Udoskonalanie narzędzi genetycznych, takich jak promotory, może w przyszłości pozwolić na podobnie efektywną produkcję innych związków terapeutycznych. To szansa dla przemysłu wielkoskalowego oraz, w tym przypadku, dla rozwoju biologicznych środków ochrony roślin.


# Część praktyczna: API krok po kroku

Podmieńcie klucz w następnej komórce. Reszta działa bez zmian.


## 5.1 Konfiguracja

Jedyna linia do podmiany to `KLUCZ`. Nagłówek `User-Agent` jest
potrzebny, bo klient korzysta z `urllib`.


````python
KLUCZ = "YOUR_KEY" # CHANGE IT
URL   = "https://hyppe.futura.foundation"

import json, statistics, urllib.error, urllib.request

UA = ("Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 "
      "(KHTML, like Gecko) Chrome/128.0 Safari/537.36")


def wolaj(sciezka, dane=None, limit=900, prob=6):
    """(kod_http, odpowiedz). `dane` != None -> POST z JSON-em.
    """
    import time
    for nr in range(prob):
        z = urllib.request.Request(URL.rstrip('/') + sciezka)
        z.add_header('X-API-Key', KLUCZ)
        z.add_header('User-Agent', UA)
        if dane is not None:
            z.add_header('Content-Type', 'application/json')
            z.data = json.dumps(dane).encode()
        try:
            with urllib.request.urlopen(z, timeout=limit) as o:
                return o.status, json.loads(o.read().decode())
        except urllib.error.HTTPError as e:
            tresc = e.read().decode('utf-8', 'replace')
            ponawialne = (e.code == 503
                          or (e.code == 429 and sciezka != '/wgraj'))
            if not ponawialne or nr == prob - 1:
                return e.code, tresc
            czekaj = float(e.headers.get('Retry-After') or 0) or 0.4 * 2 ** nr
            time.sleep(min(czekaj, 8.0))
        except urllib.error.URLError as e:
            if nr == prob - 1:
                return 0, 'siec: %s' % e
            time.sleep(0.4 * 2 ** nr)


kod_, ja = wolaj('/me')
assert kod_ == 200, ('zly klucz albo brak User-Agent', ja)
print('druzyna    :', ja['druzyna'], '|', ja['uczestnik'])
print('klucz wazny:', ja['wazny_do'], '(%.1f h)' % (ja['wazny_jeszcze_s'] / 3600)
      if ja.get('wazny_jeszcze_s') else 'klucz bez terminu')
print('limity/min :', ja['limity_na_minute'])
print('wgranie za :', ja['zgloszenie_mozliwe_za_s'], 's')
````

## 5.3 Nawigator: mapa sekwencji

`POST /nawigator/mapa` opisuje każdą pozycję sekwencji.

| kolumna | znaczenie |
|---|---|
| `rekon` | `1` = pozycja odtwarza się z samych kodów, `0` = swobodna |
| `L1 L2 L3` | `1` = zmiana jednego kodu tego poziomu rusza tę pozycję |
| `zmien_na` | zasada, na którą zmienić wejście pod P1; `.` = bez zmiany |
| `wagaP` | znormalizowana wielkość gradientu|

Pozycja z `rekon = 1` i zerami we wszystkich warstwach zostanie
nadpisana przez dekoder. Klucz `kompakt` zawiera te same dane jako
równoległe tablice, gotowe dla `pandas.DataFrame`.


````python
kod_, m = wolaj('/nawigator/mapa', {'sekwencja': DZIKI, 'od': 0, 'ile': 800})
assert kod_ == 200, m
print('gatunek      :', m['gatunek'])
print('odtwarza sie : %.4f (nie odtwarza %d z 800)'
      % (m['rekon_frakcja'], m['nie_rekonstruuje']))
print('zmian pod P1 :', m['zmian_pod_gatunek'],
      '| blad odtworzenia:', m['blad_odtworzenia'], '<- to NIE porady')
print('warstwy      :', m['rozklad_warstw'])

rek      = [w for w in m['pozycje'] if w['zmien_na'] != '.']
swobodne = [w['poz'] for w in m['pozycje'] if w['rekon'] == 0]
nadpisze = [w['poz'] for w in m['pozycje']
            if w['rekon'] == 1 and sum(w['warstwy']) == 0]
print()
print('REKOMENDACJE pod %s: %d' % (m['gatunek'], len(rek)))
for w in rek:
    print('   pozycja %3d: %s -> %s  (warstwy %s)'
          % (w['poz'], w['wej'], w['zmien_na'], w['warstwy']))
print('SWOBODNE %d, pierwsze 15: %s' % (len(swobodne), swobodne[:15]))
print('NADPISZE %d, pierwsze 15: %s' % (len(nadpisze), nadpisze[:15]))

# import pandas as pd; pd.DataFrame(m['kompakt']).head(20)
````

## 5.6 Nawigator: propozycje edycji

`POST /nawigator/edycje` zmienia kody latentu i zwraca powstałe warianty.
Pole `zmiany` liczone jest wobec linii bazowej, czyli dekodowania bez
zmiany kodów.

| poziom | warstwa | slotów | bp/slot | alfabet |
|---|---|---|---|---|
| 0 | L1 | 50 | 16 | 4 |
| 1 | L2 | 200 | 4 | 8 |
| 2 | L3 | 400 | 2 | 4 |


````python
KANDYDATOW = 100     # CELUJEMY W 100 -- tyle ocenia Wyrocznia

kod_, e = wolaj('/nawigator/edycje', {'sekwencja': DZIKI, 'poziom': 2,
                                      'ile_kodow': 8, 'opcji': 8})
assert kod_ == 200, e
print('warstwa %s | slotow %d | alfabet %d | blad rekonstrukcji %d'
      % (e['warstwa'], e['slotow'], e['alfabet'],
         e['blad_rekonstrukcji_pozycji']))

pula = {kandydat: 'z_mapy'}
for o in e['opcje']:
    pula[o['sekwencja']] = 'edycje_%d' % o['nr']

# Kilka rund, zeby zebrac wieksza pule -- limit Nawigatora to 600/min.
for runda in range(2, 30):
    if len(pula) >= KANDYDATOW:
        break
    kod_, e2 = wolaj('/nawigator/edycje',
                     {'sekwencja': DZIKI, 'poziom': 2, 'ile_kodow': 6 + runda,
                      'opcji': 8, 'ziarno': 1000 + runda})
    if kod_ != 200:
        print('runda %d: HTTP %s' % (runda, kod_)); break
    for o in e2['opcje']:
        pula[o['sekwencja']] = 'r%d_%d' % (runda, o['nr'])
print('pula unikalnych kandydatow:', len(pula))
````

## 5.7 Porównanie wariantów z sekwencją wyjściową

Sędzia nie ma odstępu między wywołaniami, więc można sprawdzić
dowolnie wiele par w granicach limitu na minutę.


````python
wygrane = []
for i, (seq, etyk) in enumerate(list(pula.items())[:25], 1):
    kod_, s = wolaj('/sedzia', {'a': DZIKI, 'b': seq,
                                'nazwa_a': 'dziki', 'nazwa_b': etyk})
    if kod_ != 200:
        print('%s: HTTP %s' % (etyk, kod_)); break
    if s['silniejsza_idx'] == 1:
        wygrane.append((etyk, seq))
print('przebily dzikiego: %d z %d sprawdzonych (%.0f%%)'
      % (len(wygrane), i, 100.0 * len(wygrane) / max(i, 1)))
print('UWAGA: wysylamy CALA pule (%d), nie tylko zwyciezcow.' % len(pula))
for etyk, _ in wygrane[:10]:
    print('  ', etyk)
````

## 5.8 Wgranie pliku FASTA

`POST /wgraj`, raz na 5 minut. Odpowiedź podaje statystyki
filtrowania i liczbę sekwencji, które weszły do oceny.


````python
# Zwyciezcy Sedziego na poczatku pliku, potem reszta puli.
# Oceniane jest pierwsze 100 sekwencji.
LIMIT = 100
zwyciezcy = {s_ for _, s_ in wygrane}
do_wgrania = ([(e_, s_) for e_, s_ in wygrane]
              + [(e_, s_) for s_, e_ in pula.items() if s_ not in zwyciezcy])
if len(do_wgrania) > LIMIT:
    print('pula %d -> przycieta do %d (limit regulaminowy)'
          % (len(do_wgrania), LIMIT))
    do_wgrania = do_wgrania[:LIMIT]
linie = []
for etyk, seq in do_wgrania:
    linie += ['>' + etyk, seq]
fasta = '\n'.join(linie)
print('plik: %d sekwencji, %d znakow' % (len(do_wgrania), len(fasta)))

kod_, r = wolaj('/wgraj', {'fasta': fasta})
print(kod_)
if kod_ == 200:
    print('--- filtrowanie ---')
    for k2, v in r['filtrowanie'].items():
        print('   %-26s %s' % (k2, v))
    print('--- wynik ---')
    for k2 in ('ocenionych', 'dzielnik_top10', 'dzielnik_top100',
               'pozycja_top10', 'pozycja_top100', 'punkty_razem'):
        print('   %-26s %s' % (k2, r.get(k2)))
else:
    print(r)
````
